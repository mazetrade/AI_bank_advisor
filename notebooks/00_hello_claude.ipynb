```python
# Load the API key

from dotenv import load_dotenv

load_dotenv()
```

```python
# Check the key without displaying it 

import os

api_key = os.getenv("ANTHROPIC_API_KEY")
print("Key loaded:", api_key is not None)
print("Looks like an Anthropic key:", api_key is not None and api_key.startswith("sk-ant-"))
```

```python
# Create the model 

from langchain_anthropic import ChatAnthropic

llm = ChatAnthropic(model="claude-haiku-5-5")
```

```python
# The first call 

response = llm.invoke("In one sentence, what is Retrieval-Augmented Generation?")
print(response.content)
```

```python
# Look inside the response 

print(type(response))
print(response.usage_metadata)
```
